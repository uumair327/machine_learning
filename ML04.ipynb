```python
import pandas as pd
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.metrics import mean_squared_error, r2_score
```

```python
# Load the dataset
data = pd.read_csv('/content/sample_data/employee_data.csv')
```

```python
data.head()
```

```python
# Check the unique values in the original columns before encoding
print("Gender unique values:", data['Gender'].unique())
print("Position unique values:", data['Position'].unique())
```

```python
# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.3, random_state=42)

# Initialize and train the Random Forest Regressor
rf_model = RandomForestRegressor(n_estimators=100, random_state=42)
rf_model.fit(X_train, y_train)
```

```python
# Make predictions
y_pred = rf_model.predict(X_test)
```

```python
# Calculate R² score
r2 = r2_score(y_test, y_pred)
print(f"R² Score: {r2:.2f}")
```

```python
import numpy as np
from collections import Counter
from sklearn.base import BaseEstimator, ClassifierMixin
```

```python
class DecisionTree(BaseEstimator, ClassifierMixin):
    def __init__(self, max_depth=None):
        self.max_depth = max_depth

    def fit(self, X, y):
        self.classes_ = np.unique(y)
        self.tree_ = self._fit(X, y, depth=0)
        return self

    def _fit(self, X, y, depth):
        n_samples, n_features = X.shape
        unique_labels = np.unique(y)

        if len(unique_labels) == 1:
            return unique_labels[0]

        if self.max_depth is not None and depth >= self.max_depth:
            return self._most_common_label(y)

        best_split = self._find_best_split(X, y)
        if best_split is None:
            return self._most_common_label(y)

        left_X, left_y, right_X, right_y = self._split(X, y, best_split)
        left_branch = self._fit(left_X, left_y, depth + 1)
        right_branch = self._fit(right_X, right_y, depth + 1)

        return {'split': best_split, 'left': left_branch, 'right': right_branch}

    def _find_best_split(self, X, y):
        # Implement the logic to find the best split
        # For simplicity, assuming binary splits and not implementing actual split finding
        return None

    def _split(self, X, y, split):
        # Implement the logic to split the data
        return X, y, X, y

    def _most_common_label(self, y):
        counter = Counter(y)
        most_common = counter.most_common(1)
        return most_common[0][0] if most_common else None

    def predict(self, X):
        return np.array([self._predict(sample) for sample in X])

    def _predict(self, sample):
        node = self.tree_
        while isinstance(node, dict):
            if sample[node['split'][0]] <= node['split'][1]:
                node = node['left']
            else:
                node = node['right']
        return node
```

```python
from sklearn.utils import resample

class RandomForest(BaseEstimator, ClassifierMixin):
    def __init__(self, n_trees=10, max_depth=None):
        self.n_trees = n_trees
        self.max_depth = max_depth

    def fit(self, X, y):
        self.trees_ = []
        for _ in range(self.n_trees):
            X_bootstrap, y_bootstrap = resample(X, y)
            tree = DecisionTree(max_depth=self.max_depth)
            tree.fit(X_bootstrap, y_bootstrap)
            self.trees_.append(tree)
        return self

    def predict(self, X):
        predictions = np.array([tree.predict(X) for tree in self.trees_])
        return np.array([self._most_common_label(preds) for preds in predictions.T])

    def _most_common_label(self, y):
        counter = Counter(y)
        most_common = counter.most_common(1)
        return most_common[0][0] if most_common else None
```

```python
import numpy as np
from sklearn.tree import DecisionTreeClassifier
from collections import Counter
```

```python
class RandomForest:
    def __init__(self, n_trees=10, max_depth=None):
        self.n_trees = n_trees
        self.max_depth = max_depth
        self.trees = []

    def fit(self, X, y):
        self.trees = []
        for _ in range(self.n_trees):
            # Bootstrap sampling
            indices = np.random.choice(len(X), len(X), replace=True)
            X_sample = X.iloc[indices]
            y_sample = y.iloc[indices]

            # Train a decision tree
            tree = DecisionTreeClassifier(max_depth=self.max_depth)
            tree.fit(X_sample, y_sample)
            self.trees.append(tree)

    def predict(self, X):
        # Collect predictions from all trees
        predictions = np.zeros((len(X), self.n_trees), dtype=int)
        for i, tree in enumerate(self.trees):
            predictions[:, i] = tree.predict(X)

        # Majority voting
        majority_votes = np.apply_along_axis(lambda x: Counter(x).most_common(1)[0][0], axis=1, arr=predictions)
        return majority_votes
```

```python
# Initialize and train the Random Forest classifier
rf = RandomForest(n_trees=10, max_depth=5)
rf.fit(X_train, y_train)
```

```python
# Predict on the test set
y_pred = rf.predict(X_test)
```

```python
# Check the shapes
print(f"Shape of y_test: {y_test.shape}")
print(f"Shape of y_pred: {y_pred.shape}")
```

**Classification Using XGBoost**

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score, classification_report
import xgboost as xgb
```

```python
# Load the dataset
data = pd.read_csv('/content/sample_data/employee_data.csv')
```

```python
# Preprocessing: Encoding categorical columns
data['Gender'] = data['Gender'].map({'F': 0, 'M': 1})  # Encode Gender
data['Position'] = data['Position'].astype('category').cat.codes  # Encode Position
```

```python
# Create a target variable 'Salary Band'
bins = [0, 50000, 100000, 150000, 200000, float('inf')]
labels = ['Low', 'Medium', 'High', 'Very High', 'Extremely High']
data['Salary Band'] = pd.cut(data['Salary'], bins=bins, labels=labels, right=False)
```

```python
# Encode the salary bands (labels)
label_encoder = LabelEncoder()
data['Salary Band Encoded'] = label_encoder.fit_transform(data['Salary Band'])
```

```python
# Features and target
X = data[['Gender', 'Experience (Years)', 'Position']]
y = data['Salary Band Encoded']
```

```python
# Split the data into training and test sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
# Initialize the XGBoost Classifier
xg_classifier = xgb.XGBClassifier(objective='multi:softmax', num_class=len(label_encoder.classes_), n_estimators=100, random_state=42)
```

```python
# Train the model
xg_classifier.fit(X_train, y_train)
```

```python
# Predict on the test set
y_pred = xg_classifier.predict(X_test)
```

```python
# Decode the predicted labels back to their original form
y_pred_labels = label_encoder.inverse_transform(y_pred)
y_test_labels = label_encoder.inverse_transform(y_test)
```

```python
# Ensure labels match the predictions
labels = label_encoder.classes_
```

```python


# Evaluate the model
accuracy = accuracy_score(y_test, y_pred)
print(f"XGBoost Classification Accuracy: {accuracy:.2f}")
print("Classification Report:")
print(classification_report(y_test_labels, y_pred_labels, labels=labels, target_names=labels))
```
