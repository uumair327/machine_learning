```python
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
df = pd.read_csv('/content/sample_data/diabetes.csv')
```

```python
import seaborn as sns
```

```python
#independent feature and dependent features
X=df.iloc[:,:-1]
y=df.Outcome
```

```python
X,y
```

```python
### train test split
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
     X, y, test_size=0.33, random_state=42)
```

```python
X_train
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import LabelEncoder
```

```python
## Postpruning
treemodel=DecisionTreeClassifier(max_depth=2)
```

```python
treemodel.fit(X_train,y_train)
```

```python
from sklearn import tree
plt.figure(figsize=(15,10))
tree.plot_tree(treemodel,filled=True)
```

```python
#prediction
y_pred=treemodel.predict(X_test)
```

```python
y_pred
```

```python
from sklearn.metrics import accuracy_score,classification_report
```

```python
score=accuracy_score(y_pred,y_test)
print(score)
```

```python
print(classification_report(y_pred,y_test))
```

CART

```python
#independent feature and dependent features
X = df.iloc[:, :-1]  # Features
y = df['Outcome']    # Target variable
```

```python
# Convert all feature values to strings
X_str = X.applymap(str)

# Convert target variable to strings
y_str = y.astype(str)
```

```python
# Flatten the feature values and combine them with the target values
flattened_features = X_str.values.flatten()
all_values = flattened_features.tolist() + y_str.tolist()
```

```python
# Initialize and fit the LabelEncoder
le = LabelEncoder()
le.fit(all_values)
```

```python
# Transform features and target variable
encoded_features = X_str.applymap(lambda x: le.transform([x])[0])
encoded_target = le.transform(y_str)
```

```python
# Display results
print("Encoded features:")
print(encoded_features.head())
print("Encoded target:")
print(encoded_target[:5])
```

```python
# Create a CART classifier
clf = DecisionTreeClassifier()

# Train the classifier on the training set
clf.fit(encoded_features, encoded_target)
```

```python
# Predict for a new instance
new_instance = ["1", "1", "1", "1", "1", "1", "1", "1"]  # New instance must match the number of features
encoded_new_instance = [le.transform([x])[0] for x in new_instance]
encoded_new_instance = [encoded_new_instance]  # Reshape for prediction

predicted_Outcome = clf.predict(encoded_new_instance)
decoded_predicted_Outcome = le.inverse_transform(predicted_Outcome)

print("Predicted type:", decoded_predicted_Outcome[0])
```
