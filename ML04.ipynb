Random forest: Classification

Dataset : https://www.kaggle.com/datasets/abhayayare/employee-data-simulation-it-industry?resource=download

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report
```

```python
# Load the dataset
data = pd.read_csv('employee_data.csv')
```

```python
# Create a target variable 'Salary Band'
bins = [0, 50000, 100000, 150000, 200000, float('inf')]
labels = ['Low', 'Medium', 'High', 'Very High', 'Extremely High']
data['Salary Band'] = pd.cut(data['Salary'], bins=bins, labels=labels, right=False)
```

```python
# Encode categorical columns
label_encoder_gender = LabelEncoder()
label_encoder_position = LabelEncoder()
label_encoder_salary_band = LabelEncoder()
```

```python
data['Gender'] = label_encoder_gender.fit_transform(data['Gender'])
data['Position'] = label_encoder_position.fit_transform(data['Position'])
data['Salary Band'] = label_encoder_salary_band.fit_transform(data['Salary Band'])
```

```python
# Define features and target variable
X = data[['Gender', 'Experience (Years)', 'Position']]
y = data['Salary Band']
```

```python
# Split the data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
# Train the Random Forest Classifier
rf_classifier = RandomForestClassifier(n_estimators=100, random_state=42)
rf_classifier.fit(X_train, y_train)
```

```python
# Make predictions
y_pred = rf_classifier.predict(X_test)
```

```python
# Evaluate the model
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy:.2f}")
print("Classification Report:")
print(classification_report(y_test, y_pred, target_names=label_encoder_salary_band.classes_))
```
